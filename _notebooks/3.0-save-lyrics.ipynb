```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../data/raw/100canciones_rock_argentino.csv")
print(f"{len(df)} rows")
df.head(3)
```

```python
# Modify column names: lower case, replace spaces and hyphens with underscores, and remove non-alphanumeric characters
df.columns = (
    df.columns.str.lower()  # Convert to lowercase
    .str.replace(" ", "_")  # Replace spaces with underscores
    .str.replace("-", "_")  # Replace hyphens with underscores
    .str.replace(r"[^a-zA-Z0-9áéíóú'ÁÉÍÓÚÑ_]", "", regex=True)
)  # Remove non-alphanumeric characters
df.columns
```

```python
def clean_lyrics(df):
    # Create a copy to avoid modifying the original DataFrame
    df_clean = df.copy()

    # Function to process each lyrics entry
    def process_lyrics(text):
        # Check if the value is NaN
        if pd.isna(text):
            return text  # Keep NaN values as they are
        else:
            # For non-NaN values, split at "Contributors" and take everything after it
            if "Contributors" in text:
                return text.split("Contributors", 1)[-1]
            else:
                return text  # Return original text if "Contributors" not found

    # Apply the function to the lyrics column
    df_clean["lyrics_clean"] = df_clean["lyrics"].apply(process_lyrics)

    return df_clean


df = clean_lyrics(df)
df.head(3)
```

```python
# Function to clean filenames
def clean_filename(filename):
    # Convert to lowercase
    filename = filename.lower()
    # Replace spaces with underscores
    filename = re.sub(r"\s+", "_", filename)
    # Remove any invalid filename characters
    filename = re.sub(r"[^\w\-_.]", "", filename)
    return filename


# Iterate through the DataFrame
for index, row in df.iterrows():
    try:
        # Get the lyrics, título, and artista
        lyrics = row["lyrics"]
        titulo = row["título"]
        artista = row["artista"]

        # Create and clean filename
        filename = f"{titulo}_{artista}"
        filename = clean_filename(filename)

        # Save lyrics to file
        filepath = f"../outputs/{filename}.txt"
        with open(filepath, "w", encoding="utf-8") as f:
            f.write(lyrics)

        print(f"Saved: {filepath}")

    except Exception as e:
        print(f"Error processing row {index}: {e}")
```
